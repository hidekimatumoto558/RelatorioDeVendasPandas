```python
df_analise = df_Vendas
```

```python
df_t = df_analise.dropna()
display(df_t.info())
```

```python
df_t = df_analise.fillna("0").sample(15)
display(df_t)
```

```python
df_analise["Data_Venda"] = pd.to_datetime(df_analise["Data_Venda"] , format="%Y-%m-%d")
df_analise.info()
```

```python
df_analise[["Categoria" , "Produto"]] = df_analise[["Categoria" , "Produto"]].apply(lambda x: x.str.strip().str.title())
```

```python
df_analise["Valor_Unitario"] = (
    df_analise["Valor_Unitario"]
      .str.replace("R$", "", regex=False)
    .str.replace(".", "", regex=False)
    .str.replace(",", ".", regex=False)
    .astype(float)
)
```

OPerações com o Pandas

```python
df_analise["faturamento"] = df_analise["Valor_Unitario"] * df_analise["Qtd"]
display(df_analise.sample(20))
```

```python
df_analise["Qtd"] = df_analise["Qtd"].fillna(0)
```

```python
display(df_analise["Qtd"])
```

```python
display(sorted(df_analise["Data_Venda"].unique()))
display(df_analise["Categoria"].unique())
```

```python
df_analise["Status"] = np.where(df_analise["Categoria"] == "Outros" ,"Incompleto" , "Completo" )
display(df_analise.sample(20))
```

```python
display(df_analise.isna().sum())
```

```python
display(df_analise["Status"].value_counts())
```

Filtrando base de dados 

```python
df_analise["Categoria"] = df_analise["Categoria"].fillna("Outros")
display(df_analise[["Categoria"]].sample(15))

```

```python
display(df_analise["Categoria"].unique())
```

```python
display(df_analise[["Categoria"]].drop_duplicates())
```

Mapeando colunas

```python
regras_setor = {
    "Eletrônicos": "Tecnologia",
    "Periféricos": "Tecnologia",
    "Acessórios": "Diversos",
    "Móveis": "Decoração e Casa",
    "Outros": "Diversos"
}

df_analise["Setor"] = df_analise["Categoria"].map(regras_setor)
```

```python
display(df_analise)
```

```python

df_analise = df_analise.reset_index(drop= True)
```

```python
Eletrônicos = df_analise[df_analise["Categoria"]== "Eletrônicos"]
Acessórios = df_analise[df_analise["Categoria"]== "Acessórios"]
Moveis = df_analise[df_analise["Categoria"]== "Móveis"]
Periféricos = df_analise[df_analise["Categoria"]== "Periféricos"]
Outros = df_analise[df_analise["Categoria"]== "Outros"]

display(Eletrônicos.head())
display(Acessórios.head())
display(Moveis.head())
display(Periféricos.head())
display(Outros.head())
```

Filtrando com condições

```python
faturamentoGrande = df_analise[df_analise["faturamento"]>=3000]
display(faturamentoGrande)
```

```python
cond1 = df_analise["Data_Venda"]>="2026-01-25"
cond2 = df_analise["faturamento"]>= 1000
df_data = df_analise[cond1 & cond2]
display(df_data)
```

Analise com Agrupamento


```python
categoria_faturamento = df_analise[["Categoria" , "faturamento"]].groupby("Categoria").sum()
categoria_faturamento = categoria_faturamento.sort_values("faturamento" , ascending= False)
categoria_faturamento = categoria_faturamento.rename(columns= {"faturamento": "valor_total"})

display(categoria_faturamento)
```

analise de agrupamento com 3 variaveis


```python
categoria_prod_fat = df_analise[["Categoria" , "Produto" , "faturamento"]].groupby(["Categoria" , "Produto"],as_index= False ).sum()
categoria_prod_fat = categoria_prod_fat.sort_values( ["Categoria", "faturamento"] , ascending= False)



display(categoria_prod_fat)
```

usando Joins no Pandas

```python
df_meta = df_analise[df_analise["Data_Venda"].dt.month == 3]
df_meta = df_meta[["ID_Loja" , "faturamento"]].groupby("ID_Loja" , as_index=False).sum()
df_meta = df_meta.merge(df_Gerente , on= "ID_Loja" , how="left" )
df_meta["Meta_comprida?"] = np.where(df_meta["faturamento"]>= df_meta["Meta_Mensal"] , "Sim" , "Não")
display(df_meta)
```

```python
df_analise["mes_ano"] = df_analise["Data_Venda"].dt.to_period("M")
Mes_ano = df_analise[["mes_ano" , "faturamento"]].groupby("mes_ano").sum()
display(Mes_ano)
```
